Tu tarea es modificar el programa que escribiste la clase anterior, para que realice las mismas tareas, pero utilizando funciones que devuelvan valores cuando sea posible. Recordá que el programa te permitía:

● `Agregar productos`: Cada producto debe tener un nombre y un precio.

● `Consultar productos`: Muestra todos los productos en la lista junto con sus precios.

● `Eliminar productos`: A partir de su nombre.

● `Menú interactivo`: Debe ofrecer un menú para que se pueda elegir qué acción realizar.

```python
# Crear una lista vacía para almacenar el nombre de los productos y su precio.
lista_productos = []

#------------------------------------------------------------------------------------
# Función para agregar productos.
def agregar_productos():
  """
  Agrega un producto en la lista de productos.

  Dato de entrada:
    Se pide el nombre y el precio del producto al usuario mediante un input.

  Retorna:
    El "nombre del producto" y su "precio" como una tupla.
  """
  nombre_producto = input("Ingresá el nombre del producto: ").capitalize().strip()
  precio_producto = (input("Ingresá su precio sin puntos, guiones ni comas: ")).strip()
  return nombre_producto, precio_producto
#------------------------------------------------------------------------------------
# Función para consultar(mostrar) todos los productos.
def consultar_productos():
  """
  Retorna la lista de todos los productos.

  Retorna:
    Si la lista no está vacía:
      Toda la lista de productos.

  Retorna:
    Si la lista está vacía:
      Una lista vacia [].
  """
  if lista_productos:
    return lista_productos
  else:
    return []
#------------------------------------------------------------------------------------
# Función para eliminar un producto por su nombre.
def eliminar_producto():
  """
  Elimina un producto de la lista de productos.

  Dato de entrada:
    Se pide el nombre del producto al usuario mediante un input.

  Retorna:
    Si el producto existe:
      sublista: con el "nombre del producto" y su precio.

  Retorna:
    Si el producto no existe en la lista_productos:
      None: Ninguno/nada.
  """
  producto_a_eliminar = input("Ingresá el nombre del producto a eliminar: ").capitalize().strip()

  for producto in lista_productos:
    if producto_a_eliminar == producto[0]:
      lista_productos.remove(producto)
      return producto

  return None
#------------------------------------------------------------------------------------
# Función para mostrar el menú de opciones.
def mostrar_menu():
  """
  Muestra un menú de opciones para gestionar una lista de productos.

  Dato de entrada:
    Se pide al usuario una opción específicada en el menú mediante un input y se ejecuta una acción en
    función de lo ingresado.

  Datos de salida:
    Ejecuta distintas salidas con lo ingresado por el usuario como agregar, mostrar y eliminar productos de la lista_productos o finaliza el programa.
    La función por sí sola retorna None porque no tiene Return o sea no devuelve ningún dato explícito.
  """
  while True:
    print("\nMenú de opciones:")
    print("1. Agregar producto.")
    print("2. Mostrar productos.")
    print("3. Eliminar producto.")
    print("4. Salir")

    opcion_elegida = input("Ingrese un número del 1 al 4, según la opción requerida: ").strip()
    #----------------------------------------------------------------------------------------
    if opcion_elegida == "1":
      nombre_producto, precio_producto = agregar_productos()
      lista_productos.append([nombre_producto, precio_producto])
      print(f'\nProducto: {nombre_producto} y precio: ${precio_producto}, agregados con éxito.')
    #----------------------------------------------------------------------------------------
    elif opcion_elegida == "2":
      productos = consultar_productos()

      if productos:
        for indice, producto in enumerate(productos, start=1):
          print(f'\nProducto Nro: {indice}, Nombre: {producto[0]}, Precio: ${producto[1]}')
      else:
        print("\nNo hay productos cargados en la lista para ser consultados.")
    #----------------------------------------------------------------------------------------
    elif opcion_elegida == "3":
      producto_eliminado = eliminar_producto()

      if producto_eliminado:
        print(f'\nEl producto: {producto_eliminado[0]}, fue eliminado con éxito.')
      else:
        print("El producto no está en la lista.\n")
    #----------------------------------------------------------------------------------------
    elif opcion_elegida == "4":
      print("¡Gracias por utilizar el programa! ¡Adiós!")
      break
    else:
      print("¡Opción inválida! Ingrese una de las opciones enumeradas en el menú de opciones.\n")

#------------------------------------------------------------------------------------
# Función que activa el menú de opciones.
mostrar_menu()

# help(agregar_productos)
# help(consultar_productos)
# help(eliminar_producto)
# help(mostrar_menu)
```
